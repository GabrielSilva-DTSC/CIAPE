### **Como estruturar a governaça e a rastreabilidade dos dados presentes nos editais para viabilizar a base do MVP seja auditável?**

Com finalidade de suprir lacunas operacionais que o projeto possui, foi realizado um trabalho conjunto com *inteligências artificiais* (Claude Code e Gemini), para estruturar a coleta de dados e realizar a leitura dos editais. **informações principais e saídas contidas na pasta 'Uso documentado da I.A'**

    • Marco legal e hierarquia institucional identificados

    • Quadro de vagas localizado e total declarado anotado
    • Cada modalidade mapeada para categoria funcional, sem forçar categorias existentes

Para extrair os dados dos editais do MVP, executar: (a) foi identificado a esfera administrativa e do marco legal que rege a assistência estudantil na UFPB; (b) mapeamento da hierarquia institucional local (campus, polo, unidade) para o campo genérico `unidade_institucional`; (c) construção de um dicionário de tradução entre o termo local do auxílio e a categoria funcional padronizada (disponível na pasta 'data', planilha Dicionário', tabela 'vocabulario_controlado'); (d) validação desse dicionário sobre um edital-piloto (Edital Unificado de 2026).

### Parsing: Mineração de texto. 

---

Para criar um ID único que se fragmenta em 3 Níveis: Estado/IES/CAMPUS. Realiza-se uma mineração de texto que carece de uma definição de termos que são encontrados nos editais.
 
data_base_CIAPE_mvp: TEM cod_estado (25) e cod_ies (1) explícito
mas NÃO TEM cod_campi. O número do campus está escondido no texto da coluna 'campus_nome'(ex: "Campus IV - Litoral Norte (...)").
 
fonte_dos_dados: já TEM cod_estado, cod_ies e cod_campi explícitos
(1, 2, 3, 4). As colunas loc_latitude/loc_longitute vieram com o separador de milhar quebrado (ex: "-7.137.594" em vez de -7.137594) e por isso são lidas como texto, não como número.
 
Ou seja: é a tabela 'data_base_CIAPE_mvp' que precisa da mineração de texto (Campus IV -> 4). A tabela 'fonte_dos_dados' só precisa ter o ID remontado no mesmo padrão para permitir o merge, e ter lat/long corrigidas.

```python
# 1) Construir o mapa "romano -> cod_campi" a partir da tabela FONTE (df_editais),
#    que é quem detém o código oficial do campus.
df_editais['campus_romano'] = (
    df_editais['nome_edital']
    .str.extract(r'campus[\s\-_]*([IVXLCDM]+)', flags=re.IGNORECASE, expand=False)
    .str.upper()
)

mapa_campus = (
    df_editais[['campus_romano', 'cod_campi']]
    .drop_duplicates()
    .set_index('campus_romano')['cod_campi']
    .to_dict()
)
# mapa_campus fica algo como {'I': '1', 'II': '2', 'III': '3', 'IV': '4'}

# 2) Extrair o romano do nome do campus no df_mvp (mesmo padrão de texto)
df_mvp['campus_romano'] = (
    df_mvp['Nome do Campus']
    .str.extract(r'campus[\s\-_]*([IVXLCDM]+)', flags=re.IGNORECASE, expand=False)
    .str.upper()
)

# 3) Buscar o cod_campi na tabela fonte (lookup), em vez de recalcular
df_mvp['cod_campi'] = df_mvp['campus_romano'].map(mapa_campus)

faltantes = df_mvp[df_mvp['cod_campi'].isna()]
if not faltantes.empty:
    raise ValueError(
        f"Não consegui casar o campus com a tabela fonte em {len(faltantes)} linha(s). "
        f"Exemplos: {faltantes['Nome do Campus'].unique()[:5]}"
    )

df_mvp['cod_campi'] = df_mvp['cod_campi'].astype(int)
df_mvp.drop(columns='campus_romano', inplace=True)
```

```python
# df_mvp: colunas 'Estado', 'cod_ies' e 'cod_campi' (está int, precisa converter)
df_mvp['ID_completo'] = (
    df_mvp['Estado'] + '-' + df_mvp['cod_ies'] + '-' + df_mvp['cod_campi'].astype(str)
)

# df_editais: 'cod_estado', 'cod_ies' e 'cod_campi' já estão como str (célula 31)
df_editais['ID_completo'] = (
    df_editais['cod_estado'] + '-' + df_editais['cod_ies'] + '-' + df_editais['cod_campi']
)

df_mvp
```

### MERGE 
 

 Merge final: traz loc_latitude, loc_longitute, ano_edital (e outros metadados do edital) da tabela fonte (df_editais) para o df_mvp. A chave de junção é ID_completo + ano — só ID_completo duplicaria cada linha do df_mvp, pois o mesmo campus tem edital de 2025 e 2026 em df_editais.


```python
colunas_a_trazer = [
    'ID_completo',
    'ano_edital',
    'loc_latitude',
    'loc_longitute',
    'nome_edital',
    'link_download_edital',
]

df_mvp = df_mvp.merge(
    df_editais[colunas_a_trazer],
    left_on=['ID_completo', 'Ano de Públicação'],
    right_on=['ID_completo', 'ano_edital'],
    how='left',
)

# Checagem: linhas do df_mvp que não encontraram correspondência na fonte
sem_par = df_mvp[df_mvp['loc_latitude'].isna()]
if not sem_par.empty:
    print(f"[AVISO] {len(sem_par)} linha(s) sem correspondência (conferir ID_completo/ano).")
    print(sem_par[['ID_completo', 'Ano de Públicação', 'Nome do Campus']].drop_duplicates())
else:
    print("Todas as linhas encontraram correspondência na tabela fonte.")

df_mvp.head()
```

```python
# Q5 — Campi SEM Alojamento nem R.U.: existe subsídio financeiro equivalente?
campi = df_mvp['Nome do Campus'].unique()

mask_alojamento = df_mvp['Tipo de Apoio'].str.contains(r'Residência Universitária|RUMF|RUFET', case=False, na=False, regex=True)

mask_ru = df_mvp['Tipo de Apoio'].str.contains(r'Residência Universitária|RUMF|RUFET', case=False, na=False, regex=True)
tem_infra = (df_mvp[mask_alojamento | mask_ru]
    .groupby('Nome do Campus')
    .size()
)
campi_sem_infra = [c for c in campi if tem_infra.get(c, 0) == 0]

q5_subsidio_alternativo = (
    df_mvp[
        df_mvp['Nome do Campus'].isin(campi_sem_infra)
        & (df_mvp['Valor destinado ao beneficiário (R$)'] > 0)
    ]
    [['Nome do Campus', 'Tipo de Apoio', 'Valor destinado ao beneficiário (R$)']]
    .drop_duplicates()
)
# Na UFPB, todos os 4 campi têm alojamento e RU, então campi_sem_infra deve vir vazio — o código fica pronto para quando entrarem IES sem essa estrutura.
```

```python
# Q7, Q8, Q9 — média/mediana/mín/máx do valor do auxílio, por Campus / IES / Estado, por ano

def resumo_estatistico(df, coluna_nivel, coluna_metrica, incluir_ano=True):
    chave = [coluna_nivel, 'Ano de Públicação'] if incluir_ano else [coluna_nivel]
    return (
        df.groupby(chave)[coluna_metrica]
        .agg(media='mean', mediana='median', minimo='min', maximo='max', qtd_registros='count')
        .reset_index()
        .sort_values(chave)
    )

resumo_valor_campus = resumo_estatistico(df_mvp, 'Nome do Campus', 'Valor destinado ao beneficiário (R$)')
resumo_valor_ies    = resumo_estatistico(df_mvp, 'Sigla da Instiuição de Ensino', 'Valor destinado ao beneficiário (R$)')
resumo_valor_estado = resumo_estatistico(df_mvp, 'Estado', 'Valor destinado ao beneficiário (R$)')
```

```python
# Q10 — Prazo médio de inscrição, por Estado / IES / Campus
prazo_por_estado = resumo_estatistico(df_mvp, 'Estado', 'Prazo de inscrição (em dias)')
prazo_por_ies    = resumo_estatistico(df_mvp, 'Sigla da Instiuição de Ensino', 'Prazo de inscrição (em dias)')
prazo_por_campus = resumo_estatistico(df_mvp, 'Nome do Campus', 'Prazo de inscrição (em dias)')
```
